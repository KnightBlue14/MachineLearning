```python
def OllamaTalk(request):
    talk = ollama.chat(
        model = "qwen3:1.7b",
        messages =[
            {"role":"user","content":f"{request}"}
        ]
    )
    return talk["message"]["content"]

print(OllamaTalk("Summarise the content of fox.txt in one line"))
```

```python
def read_file(path):
    with open(path, 'r',encoding = "utf-8") as file:
        return file.read()
```

```python
TOOL_SCHEMAS = [
    {
        "type":"function",
        "function": {
            "name":"read_file",
            "description":"Read a text file and return contents",
            "parameters":{
                "type":"object",
                "properties":{
                    "path":{"type":"string","description":"path of the file to read"}
                },
                "required":["path"],
            },
        },
    },
]
```

```python
messages = [
    {"role":"user","content":"Summarise the content of fox.txt in one line"}
]
```

```python
response = ollama.chat(model="qwen3:1.7b", messages=messages, tools=[read_file], think=True)

messages.append(response.message)
if response.message.tool_calls:
  call = response.message.tool_calls[0]
  result = read_file(**call.function.arguments)
  messages.append({"role": "tool", "tool_name": call.function.name, "content": str(result)})

  final_response = ollama.chat(model="qwen3:1.7b", messages=messages, tools=[read_file], think=True)
  print(final_response.message.content)
```
