```python
from langchain_ollama.llms import OllamaLLM
from langchain_core.prompts import ChatPromptTemplate
from rag_vector import retriever

model = OllamaLLM(model="llama3.1")

template = """
You are an expert in answering questions about restaurants.

Here are some relevant records. Only refer to them to provide answers: {records}

Here is the question to answer: {question}
"""
prompt = ChatPromptTemplate.from_template(template)
chain = prompt | model

while True:
    print("\n\n-------------------------------")
    question = input("Ask your question (q to quit): ")
    print("\n\n")
    if question == "q":
        break
    
    records = retriever.invoke(question)
    result = chain.invoke({"records": records, "question": question})
    print(result)
```
