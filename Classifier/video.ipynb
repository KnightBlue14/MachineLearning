```python
import dlib
import cv2
import face_recognition
import torch
import datetime
```

```python
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
```

```python
cap= cv2.VideoCapture("mutashort.mp4")

frames = cap.get(cv2.CAP_PROP_FRAME_COUNT)
fps = cap.get(cv2.CAP_PROP_FPS)
```

```python
print(fps)
print(frames)
```

```python
muta_image = face_recognition.load_image_file("muta_2.png")

muta_encode = face_recognition.face_encodings(muta_image)[0]

known_encode = muta_encode

known_name = 'Mutahar'
```

```python
face_locations = []
num = 0
start = datetime.datetime.now()



while num < int(frames):
    num += 1

    ret, frame = cap.read()

    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    face_locations = face_recognition.face_locations(rgb_frame)
    face_encoding = face_recognition.face_encodings(rgb_frame,face_locations)


    
    
    for top, right, bottom, left in face_locations:
        try:
            matches = face_recognition.compare_faces(
            known_encode,
            face_encoding)
            
            if True in matches:
                first_match_index = matches.index(True)
                name = known_name
    
            cv2.rectangle(frame, (left, top), (right, bottom), (0, 0,  255), 2)
    
            cv2.putText(frame,name,(left, top - 10),cv2.FONT_HERSHEY_SIMPLEX,0.8,(0, 255, 0)
                        ,2)
     
            cv2.imwrite(f'frames/frame_{num}.jpg', frame)

        except Exception:
            pass


totaltime = datetime.datetime.now() - start

print(totaltime)
```
